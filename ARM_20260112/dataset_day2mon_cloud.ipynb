## ENA

```python
import xarray as xr
import pandas as pd
import os

# 输出目录
out_dir = "/data/ggong/ARM_monthly/ENA/mwrret2turnC1"
os.makedirs(out_dir, exist_ok=True)

# 要保存的变量
vars_to_save = ["phys_lwp", "qc_phys_lwp"]

# 月份范围
months = pd.date_range("2016-01-01", "2023-12-31", freq="MS")

for t in months:
    year = t.strftime("%Y")
    month = t.strftime("%m")
    
    # 输入文件模式
    pattern = f"/data/shared_data/ARM_data/ENA/enamwrret2turnC1/enamwrret2turnC1.c1.{year}{month}*.nc"
    
    try:
        # 打开该月所有文件
        ds = xr.open_mfdataset(pattern, combine="by_coords")
        
        # 选择变量
        ds_sel = ds[vars_to_save]
        
        # 输出文件名
        out_file = os.path.join(out_dir, f"enamwrret2turnC1.c1.{year}{month}.nc")
        
        # 保存
        ds_sel.to_netcdf(out_file)
        print(f"✔ Saved: {out_file}")
    
    except Exception as e:
        print(f"✘ Failed {year}-{month}: {e}")
```
